```python
# Creating a 1D grid 
import numpy as np 
import matplotlib.pyplot as plt 

nx = 41
length = 2.0

x = np.linspace(0, length, nx) # Create number of points (nx) equally spaced points between 0 and length
print(x) # python creates 41 points between 0 and 2
```

```python
# Grid Spacing 
dx = length / (nx - 1) # Grid spacing (dx) = the distance between grid points
print(dx)   # so every grid point separated by 0.05
```

```python
# Flow variable u

# u is a transported quantity in our first CFD problem
# For now, think of u as a wave shape
# we start with all values equal to 1

u = np.ones(nx)
print(u)
```

```python
# Now we modify the middle part of the u and make it equal to 2 
u[int(0.5 / dx):int(1.0 / dx + 1)] = 2  # means u[10:21] = 2
print(u)
```

```python
plt.plot(x,u)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Initial Condition")
plt.grid()
plt.show()
```

```python
# This is the complete version of the code 

import numpy as np 
import matplotlib.pyplot as plt 

nx = 41

length = 2.0
dx = length / (nx - 1)

x = np.linspace(0, length, nx)

u = np.ones(nx)
u[10:21] = 2  # u[int(0.5 / dx):int(1.0 / dx + 1)] = 2

plt.plot(x,u)
plt.xlabel("x")
plt.ylabel('u')
plt.title("Initial Condition for 1D linear Convection")
plt.grid()
plt.show()
```
